```python
from rdflib import Graph
```

```python
g1 = Graph().parse("mapping_aop_mesh_old.rdf")
g2 = Graph().parse("aop_mesh_mappings.rdf")

if g1.isomorphic(g2):
    print("Graphs are semantically equivalent")
else:
    print("Graphs differ")
```

```python
def rdf_diff(path1: str, path2: str) -> dict:
    g1 = Graph().parse(path1)
    g2 = Graph().parse(path2)

    return {
        "equivalent": g1.isomorphic(g2),
        "only_in_1":  list(g1 - g2),
        "only_in_2":  list(g2 - g1),
        "in_common":  list(g1 & g2),
    }

diff = rdf_diff("mapping_aop_mesh_old.rdf", "aop_mesh_mappings.rdf")
print("Equivalent:", diff["equivalent"])
print("Removed:   ", len(diff["only_in_1"]))
print("Added:     ", len(diff["only_in_2"]))
print("Common:    ", len(diff["in_common"]))
```
